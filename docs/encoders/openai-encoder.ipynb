# Using OpenAIEncoder

`OpenAIEncoder` embeds text with OpenAI's embedding models. Here we use it on its own, to embed a few documents. To route with it, pass it to a router: `IndexedRouter(encoder=encoder, routes=routes)`, as in [OpenAI Embed 3 models](https://github.com/aurelio-labs/semantic-router/blob/main/docs/encoders/openai-embed-3.ipynb).

## Getting Started

We start by installing semantic-router. `OpenAIEncoder` is part of the base install:

```python
!pip install -qU "semantic-router>=1.0.0"
```

```python
import os
from getpass import getpass
from semantic_router.encoders import OpenAIEncoder

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY") or getpass(
    "Enter OpenAI API Key: "
)

encoder = OpenAIEncoder(name="text-embedding-3-small")
```

Now we define some documents to embed. One of them is far longer than the model's token limit:

```python
docs = [
    "Semantic Router is a superfast decision-making layer for LLMs and agents.",
    "It routes requests using semantic vector space, not slow LLM generations.",
    "A very long document. " * 5000,
]
```

With `truncate=True` (the default), `OpenAIEncoder` cuts each document down to the model's token limit before sending it, so long documents don't fail. With `truncate=False`, a document over the limit raises an error instead.

```python
embeds = encoder(docs=docs, truncate=True)
len(embeds), len(embeds[0])
```

We get one 1536-dimensional vector per document.
